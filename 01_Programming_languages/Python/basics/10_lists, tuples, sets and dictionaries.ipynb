# list Comprehensions

```python
# python sequences - list, range, string, tuple

# list = [1, 2, 3]
# new_list = [new_item for item in list]
# where,
# new_item = new_list.append(new_item)
# item = variable name for each item
# list = list name

numbers = [1, 2, 3]
new_list = [n+1 for n in numbers]
print(new_list) 

# same as
new_list = []
for n in numbers:
    add_1 = n+1
    new_list.append(add_1)
print(new_list)

name = "Angela"
list = [letter for letter in name]
print(list)

# same as
list = []
for letter in name:
    list.append(letter)
print(list)

# Comprehensions for range
num1 = [i+i for i in range(1,5)]
print(num1)

# same as
num = []
for i in range(1,5):
    num.append(i+i)
print(num)

# Conditional list comprehension
# new_list = [new_item for item in list if test]

names = ["alex", "beth", "Caroline", "dave", "Eleanor", "Freddie"]
short_names = [name for name in names if len(name)<5]
print(short_names)

Long_names = [name.upper() for name in names if len(name)>5]
print(Long_names)
```
