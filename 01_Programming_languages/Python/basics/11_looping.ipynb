# For loop

```python
#Syntax
#for item in list_of_items:
#    # do some thing to each item
```

```python
primes = [2,3,5,7]

#assign's variable name prime to each and every item in primes list(one at a time) and then follows the code(here print)
for prime in primes:
    print(f"{prime} is a prime no.")
```

```python
fruits = ["Apple", "peach", "pear"]
for fruit in fruits:
    print(fruit)
    print(fruit + " Pie")
    print(fruits)
print(fruits)
```

```python
nums = [1 ,2 ,3 ,4 ,5 ]

for num in nums:
      if num == 3:
        print('Found')
        break
#         continue
        print(num)
```

```python
for num in nums :
      for letter in 'abc' :
            print(num, letter)
```

# Range function

```python
# syntax
# for number in range(start, stop, step):
#     print(number)
```

```python
for i in range(5):
    print(i)
```

```python
# numbers from 10 to 19(not including 10) 
for number in range(10, 20, 2): 
    print(number)
```

```python
# reverse
for number in range(20, 10, -2): 
    print(number)
```

```python
total = 0
for number in range(1, 101):
    total += number 
print(total)
```

```python
#print the numbers divisible by 7 and not divisible by 5,between 2000 and 3000.
for i in range(2000, 3000):
    if (i%7 == 0 and i%5 != 0):
        print(i)
```
