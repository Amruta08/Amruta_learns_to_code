# Scope

```python
#  ************ Scope demonstration *********** #

enemies = 1

def increase_enemies():
    enemies = 2
    print(f"enemies inside function: {enemies}")     # enemies inside function: 2

increase_enemies()
print(f"enemies outside function: {enemies}")       # enemies outside function: 1
```

```python
#  ************ Local Scope  ************** #

# local scope - variable is only accessible inside the function
def drink_portion():
    player_strength = 20

print(player_strength)     #--> throws name error
```

```python
#  ************ Global Scope  ************** #

# global scope -  variable is accessible throughout the code
player_health = 10

def drink_portion():
    print(player_health)
    
drink_portion()               #10
```

```python
player_health = 10
def game():
    def drink_portion():         # here, drink_function is only accesable inside game() function
        print(player_health)
    drink_portion()

drink_portion()
game() 
```

```python
# ***********  Modifying global scope  *********** 
# not a good practise though use return to save in a variable and change instead

enem = 2

def inc_ene():
    global enem  # expliclity saying a global variable named enem was defined 
    enem += 2
    print(enem)  # 4
    
inc_ene()    
print(enem)  # 4
```

```python
# ***********  Using return instead of  global scope  *********** 
ene = 2

def inc_ene():
    print(ene)     # 2
    return ene + 2
    
enemy = inc_ene()    
print(enemy)      # 4
```

```python
# ***********  Global Constants *********** 
# global constants (use uppercase to create global constants)

PI = 3.14156

def cal():
    print(PI)

cal()         #3.14156
```
