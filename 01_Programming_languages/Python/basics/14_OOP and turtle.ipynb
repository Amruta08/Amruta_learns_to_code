# Building our own classes 

```python
# creating class - we use class keyword

# class name:
# class car:

# Constructor (Initializing object)
# Specifies what should happen when our object is being constructed
# init function helps to create constructor

#   def __intit__(self):        self -> Actual object being initialized
#     initialize attributes


# Creating Constructor
class Car:      
    # Setting attributes self(obj) , seats (Attribute)
    def __init__(self, seats):
        self.seats = seats
    def enter_race_mode(self):
        self.seats = 2
          
        
# my_car = Car(5) 
# # same as my_car.seats= 5

# # my_car = enter_race_mode()
# print(my_car.seats)

class User:
    def __init__(self, userid, username):
        self.userid = userid
        self.username = username
        
        # providing default value ( parameter  not passed in init function)
        self.followers = 0
        self.following = 0
        
    def follow(self, user):
        user.followers += 1
        self.following += 1

    def function():
        pass         # pass keyword to leave function empty

# obj = class()
User_1 = User(10, "Angela")
User_2 = User(11, "jack")

# Creating attributes
# User_1.id = 881
# User_1.name = "Angela"
# print(User_1.name)

print(User_1.userid)
print(User_1.username)
print(User_1.followers)

# user 1 using follow mwthod
User_1.follow(User_2)

print(User_1.followers)
print(User_2.followers)

print(User_1.following)
print(User_2.following)
```

# Turtle event listeners

```python
# using screen events in turtle documentation

# function as inputs

# def function_a(parameter):
#     # do this
    
# def function_b():
#     # do this

# function_a(function_b)
# higher order functions --> functions that can work with other functions like fun_a here
# When using methods that you havn't created yourself use keyword arguments instead of positional arguments

#-----------------------------------------------------------------------------------------------------------------------------#

def add(n1, n2):
    return n1+n2

def sub(n1, n2):
    return n1-n2

def calculator(n1, n2, func):
    return func(n1, n2)

# higher order function example, here calculator is an higher order function
result = calculator(2, 3, add)
print(result)                      # 5

#-----------------------------------------------------------------------------------------------------------------------------#

from turtle import Turtle, Screen
tin = Turtle()
screen = Screen()

def move_forward():
    tinny.forward(10)

# start listing to events
screen.listen()

# onkey(key, function with no arguments)
# When space key is pressed then trigger move_forward function
screen.onkey(key = "space", fun = move_forward )
screen.exitonclick()
```

# Class inheritance

```python
# Process of inheriting  methods and attributes from an existing class

# defining class Animal
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("inhale, exhale.")
    
# class Fish inherits from Animal class

class Fish(Animal):
    def __init__(self):
        # super refers to (super class Animal) and initializes everything that the animal class can do in the fish class
        super().__init__()
    
    def swim(self):
        print("moving in water")
        
    def breathe(self):
        # inherit method functionality from super class
        super().breathe()
        print("Doing this underwater")
        

# creating nemo object 
nemo = Fish()
nemo.swim()
nemo.breathe()
print(nemo.num_eyes)
```
