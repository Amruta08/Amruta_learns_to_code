```python
# python in-built library to work with csv files
import csv

with open("C:\\Users\\Amruta\\.vscode\\python\\weather_data.csv") as data_file:
    
    # creates csv reader object
    data = csv.reader(data_file)
    print(data)                       #<_csv.reader object at 0x000001E76DCA7400>
```

```python
# extracting only data under temp column
with open("C:\\Users\\Amruta\\.vscode\\python\\weather_data.csv") as data_file:
    
    data = csv.reader(data_file) 
    temp = []
    
    # object  can be looped
#     for row in data:
#         print(row)
        
    for row in data:
        if row[1] != "temp":
            temp.append(int(row[1]))     
    print(temp)
```
