# Tkinter

```python
import tkinter

# Creating object
window = tkinter.Tk()

# Changing title
window.title("My first GUI Program")

# Changing size of window
window.minsize(width=500, height=300)

# Creating label
my_label = tkinter.Label(text="I am a label", font=("Arial", 24, "bold"))  # font is a tuple
                                                            # bold arial font of size 24

    
# specify how the label will be layed out
# pack() -> places text automatically in the center of the screen
# # my_label.pack()
# my_label.pack(side="left")
# my_label.pack(expand=True)
my_label.grid()
# my_label.place(100)

# this will keep the window on the screen, must be at the very end of the program
window.mainloop()
```
