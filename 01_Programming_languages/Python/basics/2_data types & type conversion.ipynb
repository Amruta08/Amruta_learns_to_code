# Number convertion

#### python converts narrower type to wider type 

```python
#int to float
print(float(23))

#float to int
print(int(3.1))

#ints are narrower than floats
#floats are wider than ints.
```

```python
#float to complex
print(complex(3.4))

#complex to float :- type error
#print(float(3.4 + 8j))

print(3j+8)

# Floats are narrower than complex no.
# complex no are wider than floats
```
