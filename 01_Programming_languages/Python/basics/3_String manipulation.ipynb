# string manipulation

```python
print("hello'world'")
print("hello\nworld")
print("hello"+ " name")

#string subscripting
print("hello"[4])      

#string functions
print("Angela".lower())
print("Angela".count("a"))

a = "BOB".lower()
print(a.count("b"))
```

```python
s = "My lucky number is %d, what is yours?"  %7
print(s)
```

```python
# string replace
c = "Hello hello"

c = c.replace("hello","uni")   #Hello uni
print(c)

s = "Hello hello hello hello"

s = s.replace("hello","world",2) # Hello world world hello

# a string can't be replaced twice
s = s.replace("hello","uni") #Hello world world uni
print(s)

# replace also works with phrases
a = "this is a string"
a = a.replace("a string","word")  #this is word

print(a)
```
