```python
# # Describe Problem

# Describe Problem :-
# here we want to execute print() when we reach 20 but it doesn't happen

# def my_function():
#   for i in range(1, 20):
#     if i == 20:
#       print("You got it")
# my_function()


# solution :-
def my_function():
  for i in range(1, 21):
    if i == 20:
      print("You got it")
my_function()
```
