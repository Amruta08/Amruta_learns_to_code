# Randomization

```python
# import random module
import random

#generate random integer
random_int = random.randint(1,10)
print(random_int)

#generate random float
random_float = random.random()
print(random_float)

#generate random float between 0-10
print(random_int*random_float)

# you can create your own module by creating a file of name.py and write code in it and then import it in  another py file 
# using import name and access methods using name.function() or name.var for variables .
# import My_module
# print(My_module.pi)
```
