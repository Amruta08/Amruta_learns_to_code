# Turtle event listeners

```python
# using screen events in turtle documentation

# function as inputs

# def function_a(parameter):
#     # do this
    
# def function_b():
#     # do this

# function_a(function_b)
# higher order functions --> functions that can work with other functions like fun_a here
# When using methods that you havn't created yourself use keyword arguments instead of positional arguments

#-----------------------------------------------------------------------------------------------------------------------------#

def add(n1, n2):
    return n1+n2

def sub(n1, n2):
    return n1-n2

def calculator(n1, n2, func):
    return func(n1, n2)

# higher order function example, here calculator is an higher order function
result = calculator(2, 3, add)
print(result)                      # 5

#-----------------------------------------------------------------------------------------------------------------------------#

from turtle import Turtle, Screen
tin = Turtle()
screen = Screen()

def move_forward():
    tin.forward(10)

# start listing to events
screen.listen()

# onkey(key, function with no arguments)
# listen when space key in pressed and then trigger move_forward function
screen.onkey(key = "space", fun = move_forward )
screen.exitonclick()
```

# Class inheritance

```python
# Process of inheriting  methods and attributes from an existing class

# defining class Animal
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("inhale, exhale.")
    
# class Fish inherits from Animal class

class Fish(Animal):
    def __init__(self):
        # super refers to (super class Animal) and initializes everything that the animal class can do in the fish class
        super().__init__()
    
    def swim(self):
        print("moving in water")
        
    def breathe(self):
        # inherit method functionality from super class
        super().breathe()
        print("Doing this underwater")
        

# creating nemo object 
nemo = Fish()
nemo.swim()
nemo.breathe()
print(nemo.num_eyes)
```

# list Comprehensions

```python
# python sequences - list, range, string, tuple

# list = [1, 2, 3]
# new_list = [new_item for item in list]
# where,
# new_item = new_list.append(new_item)
# item = variable name for each item
# list = list name

numbers = [1, 2, 3]
new_list = [n+1 for n in numbers]
print(new_list) 

# same as
new_list = []
for n in numbers:
    add_1 = n+1
    new_list.append(add_1)
print(new_list)

name = "Angela"
list = [letter for letter in name]
print(list)

# same as
list = []
for letter in name:
    list.append(letter)
print(list)

# Comprehensions for range
num1 = [i+i for i in range(1,5)]
print(num1)

# same as
num = []
for i in range(1,5):
    num.append(i+i)
print(num)

# Conditional list comprehension
# new_list = [new_item for item in list if test]

names = ["alex", "beth", "Caroline", "dave", "Eleanor", "Freddie"]
short_names = [name for name in names if len(name)<5]
print(short_names)

Long_names = [name.upper() for name in names if len(name)>5]
print(Long_names)
```

# How to iterate over a pandas dataframe

```python
student_dict = {
    "student": ["Angela", "James", "Lily"],
    "score": [56, 76, 98]
}

# looping through dictionary
for (key, value) in student_dict.items():
    print(key)
    print(value)
    
import pandas

# Creating dataframe we use dataframe function
student_dataframe = pandas.DataFrame(student_dict)
print(student_dataframe)

# loop through a data frame
for (key, value) in student_dataframe.items():
    print(key)
    print(value)

# Loop through rows of a data frame
for (index, row) in student_dataframe.iterrows():
    print(row)

for (index, row) in student_dataframe.iterrows():
    print(row.student)
    
for (index, row) in student_dataframe.iterrows():
    if row.student == "Angela":
        print(row.score)
```

# *args: Many positional arguments 

```python
# Unlimited arguments
def pr(*arg):      
    print(arg)           # value's are stored in a tuple
    print(type(arg))

pr(5,6)
    

def add(*args):     
    sum = 0
    for n in args:
        sum += n
    print(sum)
        
add(4,6)
add(1,2,3,4,5,6)


# args can also be accesed through their index
def fyu(*args):
    print(args[2])
    
fyu(1,2,3)
```

# **kwargs : Many keyworded Arguments

```python
# Unlimited keyword arguments
def calculate(n, **kwargs):
    print(kwargs)          #values are stored in a dictionary
    
    # looping through dictionary
    # for key,value in kwargs.items():
        # print(key)
        # print(value)
        
    # accesing argument add is same as accessing add key in dict
    # print(kwargs["add"])
    
    n += kwargs["add"]
    n *= kwargs["mul"]
    print(n)
    
calculate(2,add=3, mul=5)


class Car:
    def __init__(self, **kw):
        #self.make = kw["make"]
        #self.model = kw["model"]
        
        # we can also use get function instead of previous method as it won't throw any error and return none instead
        self.make = kw.get("make")
        self.model = kw.get("model")
        
my_car = Car(make = "nissan", model = "GT-R")
print(my_car.model)

my_ca = Car(make = "nissan")
print(my_ca.model)
```
