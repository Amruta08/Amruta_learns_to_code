```python
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")

op = input("Enter operator + - x /  \n")

if(op == "+"):
    sum = float(num1) + float(num2)
    print('the sum of {0} and {1} is {2}'.format(num1,num2,sum))
elif(op =="-"):
    sub = float(num1) - float(num2)
    print('the sub of {0} and {1} is {2}'.format(num1,num2,sub))
elif(op == "x"):
    mul = float(num1)*float(num2)
    print('the mul of {0} and {1} is {2}'.format(num1,num2,mul))
elif(op == "/"):
    div = float(num1)/float(num2)
    print('the div of {0} and {1} is {2}'.format(num1,num2,div))
else:
    print("Enter proper value")
```

```python
nterms = int(input("How many terms?"))

# first two terms
n1,n2 = 0,1
count = 0

#check if the number of terms is valid
if nterms <=0:
    print("Enter +ve")
# if there is one term, return n1
elif nterms==1:
    print("Fibo upto",nterms,":")
    print(n1)
# generate fibonacci series
else:
    print("Fibbo:")
    while count<nterms:
        print(n1)
        nth = n1+n2
        # update values
        n1=n2
        n2=nth
        count += 1
```

```python
height = float(input("enter your height in m: "))
weight = float(input("enter your weight in kg: "))

BMI = round(weight/height**2)

if(BMI<18.5):
    print(f" Your BMI is {BMI} ,You are underweight")
elif(BMI<25):
    print(f" Your BMI is {BMI} ,You have a normal weight")
elif(BMI<30):
    print(f" Your BMI is {BMI} ,You are slightly overweight")
elif(BMI<35):
    print(f" Your BMI is {BMI} ,You are obese")
else:
    print(f" Your BMI is {BMI} ,You are clinically obese.")
```

# Exp 3 :- List

```python
# Creating list
lis = [10,20,30,40]
list1 = ['apple','banana','mango']
list2 = ["one","two","three"]

# list1[0],list1[2] = list1[2],list1[0]  #['mango', 'banana', 'apple']
# print(list1) 

# list indexing
print(list2[0])     #one  
print(list2[-1])    #three

# list slicing
print(lis[1:3])   #[20, 30]

print(list2[1:])    #['two', 'three']
print(list2[:1])    #['one']

# remove item from list various methods
print(lis.pop())    #40
print(lis.pop(1))   #20

lis.remove(30)  #[10]
print(lis)

del lis[0] #[]
print(lis) 


# add item in list various methods
lis.append(50)         #[50]
print(lis)

lis.extend([60,70])    #[50, 60, 70]
print(lis)

lis.insert(2,100)      #[50, 60, 100, 70]
print(lis)


# list concatenation and repetation
print( lis+ [100, 101, 102] )  #[50, 60, 100, 70, 100, 101, 102]

print(lis*3) #[50, 60, 100, 70, 50, 60, 100, 70, 50, 60, 100, 70]

# list sort function
list3 = [3,4,1,5,2,6,8,9,7]
list3.sort()                      #[1, 2, 3, 4, 5, 6, 7, 8, 9]
print(list3)

# list range function
l3 = list(range(0,5))  #[0, 1, 2, 3, 4]
print(l3)

# list constructor
a = list("qazwsx")
print(a)              #['q', 'a', 'z', 'w', 's', 'x']
```
