```python
import cv2

# read an image
img = cv2.imread("sample1.png")

# show image with what name
cv2.imshow("Logo", img)

# Save image as new file name
cv2.imwrite("PantechLogo.png",img)

# show image for specifed time
# 0 - shows the image continously
cv2.waitKey(0)

# close everything
cv2.destroyAllWindows()
```

```python
import cv2

# read an image
img = cv2.imread("sample2.jpg")

# print image properties
print(img.shape)
print(img.size)
print(img.dtype)
```

```python
import cv2

img = cv2.imread("sample2.jpg")

# convert image to greyscale
GrayImg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Save greyscaled image as GreyImage.jpg file
cv2.imwrite("GreyImage.jpg", GrayImg)

# show orginal image
cv2.imshow("Original", img)

# show greyscaled image
cv2.imshow("Greyscaled",GrayImg)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# read an image
img = cv2.imread("sample2.jpg")

# convert image to greyscale
grayimage = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

#show image properties 
print(f"{img.shape} {img.size} {img.dtype}")
print(f"{grayimage.shape} {grayimage.size} {grayimage.dtype}")

# save image with specified filename
cv2.imwrite("pythonlogo.jpg", img)
cv2.imwrite("greyscaledimg.jpg", grayimage)

# show image with title
cv2.imshow("Original", img)
cv2.imshow("greyscaledimage", grayimage)

# show image infinitely
cv2.waitKey(0)

# shut all windows
cv2.destroyAllWindows()
```

```python
import cv2
import imutils
img = cv2.imread("sample2.jpg")
Resizedimg = imutils.resize(img, width=20)
cv2.imwrite("Resizedimg.jpg", Resizedimg)
```

```python
import cv2
img = cv2.imread("sample2.jpg")
gaussianBlur = cv2.GaussianBlur(img, (21,21),0)
cv2.imwrite("GaussianblurImg.jpg", gaussianBlur)
```

```python
import cv2
img = cv2.imread("sample2.jpg")
GrayImg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
thresImg = cv2.threshold(GrayImg,200,255,cv2.THRESH_BINARY)[1]
cv2.imwrite("thresholdImg.jpg", thresImg)
```

```python
import cv2
img = cv2.imread("sample2.jpg")
newimg = cv2.rectangle(img,(10,10),(10+20,10+30),(0,255,0),2)
cv2.imwrite("imgwithRec.jpg", newimg)
```

```python
import cv2
img = cv2.imread("sample2.jpg")
newimg = cv2.putText(img,"python3",(10,20),cv2.FONT_HERSHEY_SIMPLEX,0.5,(0,0,255),2)
cv2.imwrite("imgwithtext.jpg", newimg)
```
