```python
import art
import random
from game_data import data
from IPython.display import clear_output


def generate(n):
    choice = random.randint(0,49)

    n2 = data[choice]['name']
    d2 = data[choice]['description']
    c2 = data[choice]['country']
    f2 = data[choice]['follower_count']
    
    if n == 0:
        print(f"Compare A: {n2}, a {d2}, from {c2}.")
    else:
        print(f"Against B: {n2}, a {d2}, from {c2}.")
    
    return f2
    
def generate1(choice):
    n1 = data[choice]['name']
    d1 = data[choice]['description']
    c1 = data[choice]['country']
    f1 = data[choice]['follower_count']
    
    print(f"Compare A: {n1}, a {d1}, from {c1}.")
    
    return f1

def index(f):
    
    f1 = f
    
    for i in range(0, 50):
        if f1 == data[i]['follower_count']:
            return i
    

score = 0
up_next = 0
def game():
    global score
    global up_next
    if score >= 1:
        a = index(up_next)
        print(art.logo)
        generate1(a)
        f1 = up_next
        print(art.vs)
        f2 = generate(1)
        
        if f1 == f2:
            f2 = generate(1)
            
    else:
        print(art.logo)
        f1 = generate(0)
        print(art.vs)
        f2 = generate(1)
        
        if f1 == f2:
            f2 = generate(1)
            
        
    

    ans = input("Who has more followers? Type 'A' or 'B': ").lower()

    if ans == 'a' and f1> f2 :
        score += 1
        up_next = f1
        clear_output(wait=True)
        print(f"You're right! Current score: {score}.")
        game()
    elif ans == 'b' and f2>f1:
        score += 1
        up_next = f2
        clear_output(wait=True)
        print(f"You're right! Current score: {score}.")
        game() 
    else:
        print(f"Sorry, that's wrong. final score: {score}.")
        return
    
    

game()
```
