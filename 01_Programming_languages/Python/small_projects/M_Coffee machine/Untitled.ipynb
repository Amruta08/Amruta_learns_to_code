```python
from main import *

water = resources["water"]
milk = resources["milk"]
coffee = resources["coffee"]
money = 0

def generate():
    print(f"Water: {water}ml")
    print(f"Milk: {milk}ml")
    print(f"coffee: {coffee}g")
    print(f"money: ${money}")

def machine(drink):
    price = MENU[drink]["cost"]
    global water
    global milk
    global coffee
    global money
    
    water1 = MENU[drink]["ingredients"]["water"]
    coffee1 = MENU[drink]["ingredients"]["coffee"]
    if drink == "latte" or drink =="cappuccino":
        milk1 = MENU[drink]["ingredients"]["milk"]
    
    
    print("Please insert coins.")
    q = float(input("How many quarters?: "))
    q = q*0.25
    
    d = float(input("How many dimes?: "))
    d = d*0.10
    
    n = float(input("How many nickels?: "))
    n = n*0.05
    
    p = float(input("How many pennies?: "))
    p = p*0.01
    
    if drink == "espresso" :
        if water < water1 or coffee < coffee1:
            print("Not enough resources, money refunded")
            return
    else:
        if water < water1 or milk < milk1 or coffee < coffee1:
            print("Not enough resources, money refunded")
            return
        
    
    total = q+d+n+p
    total = round(total, 2)
    
    if total >= price:
        if total > price:
            balance = total - price
            balance = round(balance, 2)
            money += price
            print(f"Here is {balance}$ in change.")
        print(f"Here is your {drink}☕. Enjoy!")
        if total == price:
            money += price
    else:
        print("Sorry that's not enough money. Money refunded.")
        return
    
    if drink == "espresso": 
        if water >= water1 and coffee >= coffee1:
            water = water - water1
            coffee = coffee - coffee1
    else:
        if  water >= water1 and coffee>=coffee1 and milk >= milk1:
            water = water - water1
            coffee = coffee - coffee1
            milk = milk - milk1
        
while(True):
    user_choice = input("What would you like? (espresso/latte/cappuccino):")
    if user_choice == "off":
        break
    elif user_choice == "report":
        generate()
    else:
        machine(user_choice)
```
