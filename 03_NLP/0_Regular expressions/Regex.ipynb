# Extract financial periods from financial reporting

```python
text = '''
The gross cost of operating lease vehicles in FY2021 Q1 was $4.85 billion.
In previous quarter i.e. fy2020 Q4 it was $3 billion.
'''
```

```python
# Extract financial period
pattern = 'FY\\d{4} Q[1-4]'
matches = re.findall(pattern, text, flags=re.IGNORECASE)  #re.IGNORECASE -> enables case insensitive pattern match
print(matches)
```

```python
# Extract financial amount
pattern = '\\$([\\d\\.]+)'
matches = re.findall(pattern, text, flags=re.IGNORECASE)  #re.IGNORECASE -> enables case insensitive pattern match
print(matches)
```

```python
# Extract both financial period and amount
pattern = 'FY(\\d{4} Q[1-4])[^\\$]+\\$([\\d\\.]+)'
matches = re.findall(pattern, text, flags=re.IGNORECASE)
print(matches)
```

```python
# re.search() -> finds only the 1st occurence
matches = re.search(pattern, text, flags=re.IGNORECASE)
print(matches.groups())
```

```python
text = '''
Tesla's gross cost of operating lease vehicles in FY2021 Q1 was $4.85 billion.
BMW's gross cost of operating vehicles in FY2021 S1 was $8 billion.
'''
```

```python
pattern = 'FY(\\d{4} [QS][1-4])[^\\$]+\\$([\\d\\.]+)'
matches = re.findall(pattern, text, flags=re.IGNORECASE)
print(matches)
```

# Defining a function to extract info from text

```python
def get_pattern_match(pattern, text):
    matches = re.findall(pattern, text)
    if matches:
        return matches[0]

def extract_personal_information(text):
    age = get_pattern_match('age (\\d+)', text)
    full_name = get_pattern_match('Born(.*)\\n', text)
    birth_date = get_pattern_match('Born.*\\n(.*)\\(age', text)
    birth_place = get_pattern_match('\\(age.*\\n(.*)', text)
    return {
        'age': int(age),
        'name': full_name.strip(),
        'birth_date': birth_date.strip(),
        'birth_place': birth_place.strip()
    }
```

```python
extract_personal_information(text)
```
