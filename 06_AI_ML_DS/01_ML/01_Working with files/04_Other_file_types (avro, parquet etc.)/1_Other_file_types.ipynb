```python
import pandas as pd
import os

data = pd.DataFrame({
    "Name": ["John Doe", "Jane Smith", "Emily Davis", "Michael Brown"],
    "Age": [28, 34, 41, 37],
    "Occupation": ["Software Engineer", "Data Scientist", "Product Manager", "UX Designer"]
})

data
```

# CSV

```python
# Convert to csv
data.to_csv("emp.csv", index=False)
```

```python
# Read data and display it
df = pd.read_csv("emp.csv")
df
```

```python
# Display name and age
print("Age :- ", df["Age"].to_list())
print("Name :- ", df["Name"].to_list())
```

```python
# Calculate avg age
df["Age"].mean()
```

```python
# File size
print("CSV size:", os.path.getsize("emp.csv"))
```

```python
# Convert to avro
from fastavro import writer, parse_schema

# Specify schema
schema = {
    "doc": "Employee data",
    "name": "Employee",
    "namespace": "example",
    "type": "record",
    "fields": [
        {"name": "Name", "type": "string"},
        {"name": "Age", "type": "int"},
        {"name": "Occupation", "type": "string"}
    ]
}

# Convert dataframe to list of records
records = data.to_dict(orient="records")

# Parse schema
parsed_schema = parse_schema(schema)

# Write to avro file
with open("emp.avro", "wb") as out:
    writer(out, parsed_schema, records)
```

```python
# Read data and display it
from fastavro import reader

with open("emp.avro", "rb") as f:
  df = pd.DataFrame(reader(f))

df
```

```python
# Display name and age
print("Age :- ", df["Age"].to_list())
print("Name :- ", df["Name"].to_list())
```

```python
# Calculate avg age
df["Age"].mean()
```

```python
# File size
print("AVRO size:", os.path.getsize("emp.avro"))
```
