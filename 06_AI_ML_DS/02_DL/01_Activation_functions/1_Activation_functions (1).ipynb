# Sigmoid

```python
import math

def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

```python
sigmoid(100)
```

```python
sigmoid(-2)
```

```python
sigmoid(0)
```

# tanh

```python
def tanh(x):
  return (math.exp(x) - math.exp(-x))/(math.exp(x) + math.exp(-x))
```

```python
tanh(-50)
```

```python
tanh(50)
```

```python
tanh(0)
```

#ReLU

```python
def relu(x):
  return max(0, x)
```

```python
relu(-3)
```

```python
relu(4)
```

#Leaky ReLU

```python
def leaky_relu(x):
  return max(0.1*x, x)
```

```python
leaky_relu(-10)
```

```python
leaky_relu(8)
```

# Softmax

```python
def softmax(vec):
  import math
  total_exp = 0
  for val in vec:
    total_exp += math.exp(val)
  return [round(math.exp(val)/total_exp, 2) for val in vec]
```

```python
softmax([1.3, 5.1, 2.2, 0.7, 1.1])
```
