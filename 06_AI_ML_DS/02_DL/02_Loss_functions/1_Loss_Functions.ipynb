```python
import numpy as np
```

```python
y_pred = np.array([1,1,0,0,1])
y_true = np.array([0.30,0.7,1,0,0.5])
```

# Mean Absolute Error

### Without numpy

```python
def mae(y_pred, y_true):
  total_error = 0
  for yp, yt in zip(y_pred, y_true):
    total_error += abs(yp-yt)
  mae = total_error/len(y_pred)
  return mae
```

```python
mae(y_pred, y_true)
```

### With numpy

```python
def mae_np(y_pred, y_true):
  return np.mean(np.abs(y_pred-y_true))
```

```python
mae_np(y_pred, y_true)
```

# Mean squared error

### Without numpy

```python
def mse(y_pred, y_true):
  total_error = 0
  for yp, yt in zip(y_pred, y_true):
    total_error += (yp-yt)**2
  mse = total_error/len(y_pred)
  return mse
```

```python
mse(y_pred, y_true)
```

### With numpy

```python
def mse_np(y_pred, y_true):
  return np.mean(np.square(y_pred - y_true))
```

```python
mse_np(y_pred, y_true)
```

# Log loss function

log of 0 is undefined, so we will consider a very small value close to zero, to obtain valid log

```python
epsilon = 1e-15
y_pred_n = [max(i,epsilon) for i in y_pred]
print(y_pred_n)
y_pred_n = [min(i, 1-epsilon) for i in y_pred_n]
print(y_pred_n)
```

```python
def log_loss(y_pred, y_true):
  epsilon = 1e-15
  y_pred_new = [max(i, epsilon) for i in y_pred]
  y_pred_new = [min(i, 1-epsilon) for i in y_pred_new]
  y_pred_new = np.array(y_pred_new)
  return -np.mean(y_true*np.log(y_pred_new)+(1-y_true)*np.log(1-y_pred_new))
```

```python
log_loss(y_pred, y_true)
```
