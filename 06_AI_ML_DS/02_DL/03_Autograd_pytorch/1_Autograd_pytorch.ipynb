# Autograd
* It is a core component of pytorch that provides automatic differentiation for tensor operations



# Examples of computing differentiation with and without autograd

## Example 1 :-  y = x^2

### Without Autograd

```python
def dy_dx(x):
  return 2*x
```

```python
dy_dx(3)
```

### With Autograd

```python
import torch
x = torch.tensor(3.0, requires_grad=True)
```

```python
y = x**2
y.backward()
```

```python
x.grad
```

## Example 2 :- z = sin(y) and y = x^2

### Without Autograd

```python
import math
def dz_dx(x):
  return 2*x*math.cos(x**2)
```

```python
dz_dx(3)
```

### With Autograd

```python
import torch
x = torch.tensor(3.0, requires_grad=True)
```

```python
y = x**2
z = torch.sin(y)
z.backward()
```

```python
x.grad
```

# Backward Pass Gradient computation for a single neuron

## Without Autograd

```python
x = torch.tensor(6.8) # Input feature
y = torch.tensor(0.0) # Output label

w = torch.tensor(1.0) # Weight
b = torch.tensor(1.0) # Bias
```

```python
# Loss function
def binary_cross_entropy(prediction, target):
  epsilon = 1e-8
  prediction = torch.clamp(prediction, epsilon, 1 - epsilon)
  return -target * torch.log(prediction) - (1 - target) * torch.log(1 - prediction)
```

```python
# Forward Pass
z = w*x + b
y_pred = torch.sigmoid(z)

# Loss
loss = binary_cross_entropy(y_pred, y)
```

```python
# Derivatives
dloss_dy_pred = (y_pred - y)/(y_pred*(1-y_pred))

dy_pred_dz = y_pred * (1 - y_pred)

dz_dw = x
dz_db = 1

dL_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dL_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
dL_dw, dL_db
```

## With Autograd

```python
x = torch.tensor(6.8) # Input feature
y = torch.tensor(0.0) # Output label

w = torch.tensor(1.0, requires_grad=True) # Weight
b = torch.tensor(1.0, requires_grad=True) # Bias
```

```python
# Forward Pass
z = w*x + b
y_pred = torch.sigmoid(z)

# Loss
loss = binary_cross_entropy(y_pred, y)
```

```python
loss.backward()
```

```python
w.grad, b.grad
```

# Disabling gradient tracking

### option 1 - requires_grad_(False)

```python
x = torch.tensor(2.0, requires_grad=True)
```

```python
y = x**2
y.backward()
```

```python
x.grad
```

```python
# Will throw runtime error
x.requires_grad_(False)
y = x**2
y.backward()
```

### option 2 - detach()


```python
x = torch.tensor(2.0, requires_grad=True)
```

```python
y = x**2
y.backward()
```

```python
x.grad
```

```python
# Will throw runtime error
y = x.detach()
y = y**2
y.backward()
```

### option 3 - torch.no_grad()

```python
x = torch.tensor(2.0, requires_grad=True)
```

```python
y = x**2
y.backward()
```

```python
x.grad
```

```python
# will throw runtime error
with torch.no_grad():
  y = x**2
  y.backward()
```
