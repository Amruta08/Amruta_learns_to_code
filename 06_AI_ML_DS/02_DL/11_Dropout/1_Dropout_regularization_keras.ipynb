# Loading data + preprocessing

```python
import pandas as pd
import numpy as np
df = pd.read_csv('sonar_dataset.csv', header=None)
df.head(5)
```

```python
# Label encoding output column
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df[60] = le.fit_transform(df[60])
df.head(3)
```

```python
from sklearn.model_selection import train_test_split
X = df.drop(columns=60, axis=1)
y = df[60]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

# Deep learning model training without dropout layer

```python
import tensorflow as tf
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(X_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(X_test, y_test)
```

Clearly, training accuracy >> test accuracy

```python
y_pred = np.round(model.predict(X_test).reshape(-1))
y_pred
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

# Deep learning model training with dropout layer

```python
import tensorflow as tf
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dropout(0.5), #0.5 means to drop 50% of neurons
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(X_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(X_test, y_test)
```

```python
y_pred = np.round(model.predict(X_test).reshape(-1))
y_pred
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```
