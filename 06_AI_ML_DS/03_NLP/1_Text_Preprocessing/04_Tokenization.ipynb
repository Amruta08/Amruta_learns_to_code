#### Extracting all money transactions

```python
transactions = "Tony gave two $ to Peter, Bruce gave 500 € to Steve"

doc = nlp(transactions)
for token in doc:
  if token.is_currency:
    prev_index = token.i-1
    print(doc[prev_index], token.text)
```

#### Extracting email ids using token attributes

```python
with open("students.txt") as f:
  text = f.readlines()
text
```

```python
text = " ".join(text)
text
```

```python
doc = nlp(text)
emails=[]
for token in doc:
  if token.like_email:
    emails.append(token.text)

print(emails)
```

#### Extracting urls from text

```python
text='''
Look for data to help you address the question. Governments are good
sources because data from public research is often freely available. Good
places to start include http://www.data.gov/, and http://www.science.
gov/, and in the United Kingdom, http://data.gov.uk/.
Two of my favorite data sets are the General Social Survey at http://www3.norc.org/gss+website/,
and the European Social Survey at http://www.europeansocialsurvey.org/.
'''

doc = nlp(text)
urls = []
for token in doc:
  if token.like_url:
    urls.append(token.text)

print(urls)
```

## Customizing tokenizer

```python
from spacy.symbols import ORTH
nlp = spacy.blank("en")

doc = nlp("gimme double cheese extra large healthy pizza")
tokens = [token.text for token in doc]
tokens
```

```python
nlp.tokenizer.add_special_case("gimme",[
    {ORTH:"gim"},
    {ORTH:"me"},
])
doc = nlp("gimme double cheese extra large healthy pizza")
tokens = [token.text for token in doc]
tokens
```

## Sentence tokenization

```python
doc = nlp("The dog jumps over the fence. The cat is sleeping.")
for sent in doc.sents:
  print(sent) # Gives error since sentence tokenizer is not in nlp pipeline
```

```python
nlp.add_pipe('sentencizer')
doc = nlp("The dog jumps over the fence. The cat is sleeping.")
for sent in doc.sents:
  print(sent)
```
