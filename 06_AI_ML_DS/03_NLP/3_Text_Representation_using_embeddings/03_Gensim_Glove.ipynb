```python
import gensim.downloader as api
model = api.load("glove-twitter-25")
```

```python
# vector representation of good
model.get_vector("good")
```

```python
# get words similar to good
model.most_similar("good")
```

```python
# Vector arithmetic
model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5)
```

```python
# odd man out
model.doesnt_match(['breakfast', 'cereal', 'dinner', 'lunch'])
```

```python
model.doesnt_match(['cherry', 'grape', 'orange', 'dog'])
```
