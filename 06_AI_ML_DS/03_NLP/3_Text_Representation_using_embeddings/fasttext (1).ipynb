```python
import fasttext
from huggingface_hub import hf_hub_download

# Download fasttext English model file
model_path = hf_hub_download(repo_id="facebook/fasttext-en-vectors", filename="model.bin")
```

```python
# Load the model
model = fasttext.load_model(model_path)
```

```python
# Number of words in vocabulary
len(model.words)
```

```python
# Vector representation for bread
model['bread']
```

```python
# Get similar words
model.get_nearest_neighbors('good')
```

```python
# Get analogies
model.get_analogies("berlin","germany","france")
```

```python
model.get_analogies("driving", "car", "book")
```

# Fasttext on indian food recipies

```python
import pandas as pd
df = pd.read_csv('Cleaned_Indian_Food_Dataset.csv')
df.sample(2)
```

```python
# Removing special characters and newlines from text
import re

def clean_text(text):
  text = re.sub(r'[^\w\s\']', ' ', text)
  text = re.sub(r'[ \n]+', ' ', text)
  return text.strip().lower()
```

```python
df['TranslatedInstructions'] = df['TranslatedInstructions'].map(clean_text)
```

```python
# Save to csv
df.to_csv("food_receipes.txt", columns=["TranslatedInstructions"], header=None, index=False)
```

```python
# Train fasttext model
import fasttext
model = fasttext.train_unsupervised("food_receipes.txt")
```

```python
model.get_nearest_neighbors("paneer")
```

```python
model.get_nearest_neighbors("chutney")
```

```python
model.get_nearest_neighbors("halwa")
```
