# Loading dataset and preprocessing

```python
import pandas as pd
df = pd.read_csv("Emotion_classify_Data.csv")
df.head(5)
```

```python
df['Emotion'].value_counts()
```

```python
# Mapping target column
df['Emotion_num'] = df['Emotion'].map({
    'anger' : 0,
    'joy': 1,
    'fear': 2,
})

df.head(5)
```

```python
# Preprocess function
import spacy
nlp = spacy.load("en_core_web_sm", disable=["parser", "tagger", "ner"])

def preprocess(texts):
    results = []
    for doc in nlp.pipe(texts, batch_size=1000):
        tokens = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
        results.append(" ".join(tokens))
    return results
```

```python
df["Comment"] = preprocess(df["Comment"])
```

```python
df.head()
```

```python
# Train test split
from sklearn.model_selection import train_test_split
X = df['Comment']
Y = df['Emotion_num']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=df['Emotion_num'])
```

# Applying BoW and Naive Bayes, Random forest

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
from sklearn.metrics import classification_report
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('rf', RandomForestClassifier())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

# Applying n-grams and Naive Bayes, Random forest

```python
clf = Pipeline([
    ('vectorizer', CountVectorizer(ngram_range=(2,2))),
    ('nb', MultinomialNB())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

```python
clf = Pipeline([
    ('vectorizer', CountVectorizer(ngram_range=(2,2))),
    ('rf', RandomForestClassifier())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

# Applying TF-IDF and Naive Bayes, Random forest

```python
from sklearn.feature_extraction.text import TfidfVectorizer
clf = Pipeline([
    ('vectorizer', TfidfVectorizer()),
    ('nb', MultinomialNB())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

```python
clf = Pipeline([
    ('vectorizer', TfidfVectorizer()),
    ('rf', RandomForestClassifier())
])

clf.fit(X_train, Y_train)
clf.score(X_test, Y_test)
```

```python
y_pred = clf.predict(X_test)
print(classification_report(Y_test, y_pred))
```

# Using fasttext vectors

```python
df['label'] = '__label__' + df['Emotion'].astype(str)
df.head()
```

```python
# Combining label and comment
df['label_comment'] = df['label'] + ' ' + df['Comment']
df.head(3)
```

```python
# Train test split
from sklearn.model_selection import train_test_split
df = df[['label_comment']]

train, test = train_test_split(df, test_size=0.2)
```

```python
train.to_csv("emotion.train", columns=["label_comment"], index=False, header=False)
test.to_csv("emotion.test", columns=["label_comment"], index=False, header=False)
```
