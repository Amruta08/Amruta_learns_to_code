```python
import numpy as np
import tensorflow as tf
```

```python
sentences = [
    "how are you",
    "how are you doing",
    "have a good day",
    "have a great day",
    "good morning",
    "good night",
    "take care",
    "see you soon",
    "thank you so much",
    "you are welcome",
    "I hope you are well",
    "what are you up to",
    "nice to meet you",
    "have fun",
    "stay safe"
]
```

### Tokenize words

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
vocab_size = len(tokenizer.word_index) + 1  # +1 to include index 0 for padding
```

### Create training pairs

```python
# Convert words to sequence
seqs = tokenizer.texts_to_sequences(sentences)
seqs
```

```python
# X = partial sentence, y = next word
X, y = [], []
for s in seqs:
  for i in range(1, len(s)):
    X.append(s[:i])
    y.append(s[i])

    print("X :- ", s[:i], "and y :-", s[i])
```

```python
print(X)
```

```python
print(y)
```

### Padding

```python
# Pad X to equal length
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad_len = max(len(i) for i in X)
X = pad_sequences(X, maxlen=pad_len)
X
```

```python
y = np.array(y)
```

### Defining model + training + prediction

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=16),
    SimpleRNN(units=32),
    Dense(vocab_size, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')

model.fit(X, y, epochs=60, verbose=0)
```

```python
# Prediciton function
def predict_next_word(text):
  test_text = text.lower()
  test_seq = tokenizer.texts_to_sequences([test_text])[0]  # Convert text to sequence
  test_seq = pad_sequences([test_seq], maxlen=pad_len)     # pad sequence
  pred_idx = np.argmax(model.predict(test_seq), axis=1)[0] # get index of next word having max prob.
  pred_word = [word for word, idx in tokenizer.word_index.items() if idx == pred_idx] # get word by index
  print("Input:", test_text)
  print("Next word prediction:", pred_word[0] if pred_word else "(unknown)")
```

```python
predict_next_word("have a good")
```

```python
predict_next_word("how are")
```

```python
predict_next_word("nice to meet")
```
