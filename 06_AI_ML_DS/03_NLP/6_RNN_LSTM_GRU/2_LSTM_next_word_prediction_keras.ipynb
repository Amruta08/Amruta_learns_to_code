```python
import numpy as np
import tensorflow as tf
```

```python
sentences = [
    "how are you doing today",
    "hope your day goes well ahead",
    "have a good day and stay safe",
    "wishing everyone a happy new year",
    "tensorflow makes building models simple",
    "LSTM remembers context for longer sequences",
    "nice to meet you and take care",
    "take care and stay safe always",
    "hello hope everything is fine",
    "good morning have a productive day",
    "good night sleep well tonight",
    "thank you for the support today",
    "you are welcome take care",
    "see you soon stay safe",
    "I hope everything is going great",
    "let me know if anything is needed",
    "it was nice talking today",
    "have fun and enjoy life",
    "learning new things takes time",
    "do not worry take it slow",
    "sometimes it feels difficult but keep going",
    "what are you doing these days",
    "appreciate the help and guidance",
    "hope we meet again soon",
    "glad we got a chance to talk",
    "this is a simple sequence learning demo",
    "LSTM handles longer context than simple RNN",
    "GRU is simpler but still remembers context",
    "the model learns patterns from word sequences",
    "take breaks drink water stay consistent",
    "hope your goals work out well ahead",
    "good to see you today",
    "keep going one step at a time",
    "hope everything works out fine",
    "stay consistent and keep learning",
    "have a productive and calm day ahead",
    "thank you take care stay safe",
    "hope the new year brings happiness",
    "deep learning works best with good patterns",
    "start small build step by step",
    "context matters more than memorizing endings",
    "keep learning and stay consistent always"
]
```

### Tokenization

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
word_index = tokenizer.word_index
print(len(word_index))
```

```python
vocab_size = len(tokenizer.word_index) + 1  # +1 to include index 0 for padding
```

```python
# Convert words to sequence
seqs = tokenizer.texts_to_sequences(sentences)
```

### Create training pairs

```python
# X = partial sentence, y = next word
X, y = [], []
for s in seqs:
  for i in range(1, len(s)):
    X.append(s[:i])
    y.append(s[i])
```

```python
y = np.array(y)
```

### Padding

```python
# Pad X to equal length
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad_len = max(len(i) for i in X)
X = pad_sequences(X, maxlen=pad_len)
```

### Defining model + training + prediction

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=16),
    LSTM(64),
    Dense(vocab_size, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')

model.fit(X, y, epochs=30, verbose=0)
```

```python
# Prediciton function
def predict_next_word(text):
  test_text = text.lower()
  test_seq = tokenizer.texts_to_sequences([test_text])[0]  # Convert text to sequence
  test_seq = pad_sequences([test_seq], maxlen=pad_len)     # pad sequence
  pred_idx = np.argmax(model.predict(test_seq), axis=1)[0] # get index of next word having max prob.
  pred_word = [word for word, idx in tokenizer.word_index.items() if idx == pred_idx] # get word by index
  print("Input:", test_text)
  print("Next word prediction:", pred_word[0] if pred_word else "(unknown)")
```

```python
predict_next_word("how are")
```
