```python
# Read image
import cv2
img = cv2.imread("bird.jpg")
img
```

```python
# Display the image in notebook (OpenCV loads images in BGR)
import matplotlib.pyplot as plt
plt.imshow(img)
```

```python
# Display the image in notebook in RGB
import matplotlib.pyplot as plt
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # Convert BGR to RGB
plt.imshow(img)
```

```python
# Write to the image by adding 30 value to each of its pixel values
img_new = img + 30
cv2.imwrite("output0.jpg", img_new)
img_new = cv2.cvtColor(img_new, cv2.COLOR_BGR2RGB) # Convert BGR to RGB
plt.imshow(img_new)
```

```python
# Read and display video
vid = cv2.VideoCapture("flower.mp4")

while True:
    ret, frame = vid.read() # Capture frame-by-frame
    if not ret: # if frame is read correctly ret is True
        break
    cv2.imshow("Video Frame", frame) # Display the frame
    if cv2.waitKey(1) == ord('q'):
        break

vid.release() # release the capture
cv2.destroyAllWindows()
```

```python
# Write a video
vid = cv2.VideoCapture("flower.mp4")

fourcc = cv2.VideoWriter_fourcc(*"mp4v") # video codec
fps = 30 # video fps
width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH)) # Get video width from video
height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT)) # Get height from video

out = cv2.VideoWriter("output0.mp4", fourcc, fps, (width, height)) # save video

while True:
    ret, frame = vid.read()
    if not ret:
        break
    frame = frame + 30  #adding 30 value to each of its pixel values
    out.write(frame)

out.release()
vid.release()
cv2.destroyAllWindows()
```

```python
# Capture video from camera
cap = cv2.VideoCapture(0) # Value = 0 -> Read from camera

while True:
    ret, frame = cap.read()
    if not ret:
        break
    cv2.imshow("Camera", frame)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
