```python
import cv2
import matplotlib.pyplot as plt
img = cv2.imread('fish.jpg')
```

image.png

# Grayscale

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite('gray_fish.jpg', gray_img)
```

image.png

# HSV

```python
hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.imwrite('hsv_fish.jpg', hsv_img)
```

image.png

# LAB

```python
lab_img = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
cv2.imwrite('lab_fish.jpg', lab_img)
```

image.png

# RGB

```python
# (Wrong) Save as RGB => Saves as BGR (RGB gets flipped to BGR while saving)
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
cv2.imwrite('bgr_fish.jpg', rgb_img)
```

```python
# (Correct) Save as RGB => BGR gets flipped to RGB
cv2.imwrite('rgb_fish.jpg', img)
```
