```python
import matplotlib.pyplot as plt
import cv2

img = cv2.imread("giraffe.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
plt.imshow(img_rgb)
```

# Example 1

```python
# Create a binary mask for white regions
import numpy as np
lower_white = np.array([0, 0, 190]) #lower bound of white color
upper_white = np.array([30, 80, 255]) #upper bounf og white color
mask = cv2.inRange(img_hsv, lower_white, upper_white)
plt.imshow(mask)
```

```python
# Replace those white regions with black color
black_hsv = np.uint8([[[120, 255, 0]]]) # define black color in hsv
black_bgr = cv2.cvtColor(black_hsv, cv2.COLOR_HSV2BGR)
img_result = img.copy()
img_result[mask>0] = black_bgr
img_result = cv2.cvtColor(img_result, cv2.COLOR_BGR2RGB)
plt.imshow(img_result)
```

# Example 2

```python
# Create a binary mask for blue regions
import numpy as np
lower_blue = np.array([90, 50, 190]) #lower bound of blue color
upper_blue = np.array([130, 255, 255]) #upper bounf of blue color
mask = cv2.inRange(img_hsv, lower_blue, upper_blue)
plt.imshow(mask)
```

```python
# Replace those blue regions with orange color
orange_hsv = np.uint8([[[20, 200, 255]]]) # define orange color in hsv
orange_bgr = cv2.cvtColor(orange_hsv, cv2.COLOR_HSV2BGR)
img_result = img.copy()
img_result[mask>0] = orange_bgr
img_result = cv2.cvtColor(img_result, cv2.COLOR_BGR2RGB)
plt.imshow(img_result)
```
