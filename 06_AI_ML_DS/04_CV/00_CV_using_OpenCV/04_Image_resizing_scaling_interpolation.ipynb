```python
import matplotlib.pyplot as plt
import cv2

original_img = cv2.imread("giraffe.png")
original_img = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
original_img.shape
```

# Image resizing

```python
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title("Original Image")

plt.subplot(1, 2, 2)
width = 400
height = 400
resized_img = cv2.resize(original_img, (width, height))
plt.imshow(resized_img)
plt.title("Resized Image")
```

# Image Scaling

```python
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title(f"Original Image:- {original_img.shape}")

plt.subplot(1, 2, 2)
scale_percent = 50
s_height = int(original_img.shape[0]*scale_percent/100)
s_width = int(original_img.shape[1]*scale_percent/100)
scaled_img = cv2.resize(original_img, (s_width, s_height))
plt.imshow(scaled_img)
plt.title(f"Resized Image:- {scaled_img.shape}")
```

# Interpolation

```python
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
plt.imshow(original_img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
img1 = cv2.resize(original_img, (600, 600), interpolation=cv2.INTER_NEAREST)
plt.imshow(img1)
plt.title("INTER_NEAREST")
plt.axis("off")

plt.subplot(2, 3, 3)
img2 = cv2.resize(original_img, (600, 600), interpolation=cv2.INTER_LINEAR)
plt.imshow(img2)
plt.title("INTER_LINEAR")
plt.axis("off")

plt.subplot(2, 3, 4)
img3 = cv2.resize(original_img, (600, 600), interpolation=cv2.INTER_CUBIC)
plt.imshow(img3)
plt.title("INTER_CUBIC")
plt.axis("off")

plt.subplot(2, 3, 5)
img4 = cv2.resize(original_img, (600, 600), interpolation=cv2.INTER_AREA)
plt.imshow(img4)
plt.title("INTER_AREA")
plt.axis("off")

plt.subplot(2, 3, 6)
img5 = cv2.resize(original_img, (600, 600), interpolation=cv2.INTER_LANCZOS4)
plt.imshow(img5)
plt.title("INTER_LANCOZOS4")
plt.axis("off")
```
