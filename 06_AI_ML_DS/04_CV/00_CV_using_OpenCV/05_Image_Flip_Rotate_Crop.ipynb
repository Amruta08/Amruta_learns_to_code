```python
import matplotlib.pyplot as plt
import cv2

original_img = cv2.imread("giraffe.png")
original_img = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
original_img.shape
```

# Flip



*   1 -> Horizontal axis
*   0 -> Vertical axis
*   -1 -> Both axes



```python
plt.subplot(2,2,1)
plt.imshow(original_img)
plt.title("Original image")
plt.axis("off")

plt.subplot(2,2,2)
flip_h = cv2.flip(original_img, 1)
plt.imshow(flip_h)
plt.title("Flip Horizontal")
plt.axis("off")

plt.subplot(2,2,3)
flip_v = cv2.flip(original_img, 0)
plt.imshow(flip_v)
plt.title("Flip Vertical")
plt.axis("off")

plt.subplot(2,2,4)
flip_both = cv2.flip(original_img, -1)
plt.imshow(flip_both)
plt.title("Flip Both axes")
plt.axis("off")
```

# Rotate

```python
plt.subplot(2,2,1)
plt.imshow(original_img)
plt.title("Original image")
plt.axis("off")

plt.subplot(2,2,2)
rotate_90_clockwise = cv2.rotate(original_img, cv2.ROTATE_90_CLOCKWISE)
plt.imshow(rotate_90_clockwise)
plt.title("90 degree clockwise")
plt.axis("off")

plt.subplot(2,2,3)
rotate_90_counterclockwise = cv2.rotate(original_img, cv2.ROTATE_90_COUNTERCLOCKWISE)
plt.imshow(rotate_90_counterclockwise)
plt.title("90 degree counterclockwise")
plt.axis("off")

plt.subplot(2,2,4)
rotate_180 = cv2.rotate(original_img, cv2.ROTATE_180)
plt.imshow(rotate_180)
plt.title("180 degrees")
plt.axis("off")
```

# Crop

```python
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title("Original Image")

plt.subplot(1, 2, 2)
x_start, y_start = 400, 300
x_end, y_end = 800, 900
cropped_img = original_img[y_start:y_end, x_start:x_end]
plt.imshow(cropped_img)
plt.title("Cropped Image")
```
