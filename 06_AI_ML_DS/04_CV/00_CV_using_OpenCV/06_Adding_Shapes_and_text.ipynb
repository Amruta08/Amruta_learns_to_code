```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

# Drawing shapes

```python
# Create an empty canvas
height, width = 500, 700
blue = (255, 127, 0)
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Draw a line
start = (50, 50)
end = (650, 450)
color = (255, 255, 255)
thickness = 5
cv2.line(canvas, start, end, color, thickness)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Draw a rectangle
start = (100, 100)
end = (300, 200)
color = (0, 255, 255)
thickness = -1 # -1 indicates to fill
cv2.rectangle(canvas, start, end, color, thickness)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Draw a circle
center = (500, 150)
radius = 100
color = (0, 0, 255)
thickness = 5
cv2.circle(canvas, center, radius, color, thickness)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Draw an Ellipse
center = (350, 350)
axes = (100, 50)
angle = 45
start_angle = 0
end_angle = 360
color = (0, 255, 0)
thickness = 3
cv2.ellipse(canvas, center, axes, angle, start_angle, end_angle, color, thickness)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Draw a polygon
points = np.array([[150, 400], [300, 400], [225, 300]], np.int32)
points = points.reshape((-1, 1, 2))
is_closed = True
color = (255, 0, 255)
thickness = 3
cv2.polylines(canvas, [points], is_closed, color, thickness)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Adding text

```python
# Create an empty canvas
height, width = 500, 700
blue = (255, 127, 0)
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Add text
text = "Hi, There!!"
point = (200, 100) #point for bottom left corner of the text
font = cv2.FONT_HERSHEY_COMPLEX
font_scale = 1.5
color = (0, 255, 255)
thickness = 2
line_type = cv2.LINE_AA
cv2.putText(canvas, text, point, font, font_scale, color, thickness, line_type)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Different font styles
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

text = "Hi, There!!"
font_scale = 1
color = (0, 255, 255)
thickness = 2
line_type = cv2.LINE_AA

cv2.putText(canvas, text, (250, 50), cv2.FONT_HERSHEY_SIMPLEX, font_scale, color, thickness, line_type)
cv2.putText(canvas, text, (250, 100), cv2.FONT_HERSHEY_COMPLEX, font_scale, color, thickness, line_type)
cv2.putText(canvas, text, (250, 150), cv2.FONT_HERSHEY_SCRIPT_SIMPLEX, font_scale, color, thickness, line_type)
cv2.putText(canvas, text, (250, 200), cv2.FONT_HERSHEY_SCRIPT_COMPLEX, font_scale, color, thickness, line_type)
cv2.putText(canvas, text, (250, 250), cv2.FONT_HERSHEY_COMPLEX_SMALL, font_scale, color, thickness, line_type)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

```python
# Font size and thickness
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

text = "Hi, There!!"
font = cv2.FONT_HERSHEY_COMPLEX
color = (0, 255, 255)
line_type = cv2.LINE_AA

cv2.putText(canvas, text, (200, 50), font, 1, color, 2, line_type)
cv2.putText(canvas, text, (200, 150), font, 2, color, 2, line_type)
cv2.putText(canvas, text, (200, 250), font, 1, color, 4, line_type)
cv2.putText(canvas, text, (200, 350), font, 2, color, 4, line_type)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```
