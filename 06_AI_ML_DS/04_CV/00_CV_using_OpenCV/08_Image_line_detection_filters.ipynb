```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

img = cv2.imread('filter.jpg')
plt.imshow(img, cmap='gray')
```

## Horizontal lines

```python
x = -10
horizontal_kernel = np.array([[-1, -1, -1],
                              [x, x, x],
                              [1, 1, 1]])

horizontal_filtered_img = cv2.filter2D(img, -1, horizontal_kernel)
plt.imshow(horizontal_filtered_img, cmap='gray')
```

## Vertical lines

```python
x = -10
vertical_kernel = np.array([[-1, x, 1],
                            [-1, x, 1],
                            [-1, x, 1]])

vertical_filtered_img = cv2.filter2D(img, -1, vertical_kernel)
plt.imshow(vertical_filtered_img, cmap='gray')
```

## Diagonal lines

```python
x = -10
diagonal1_kernel = np.array([[x, -1, -1],
                            [1, x, -1],
                            [1, 1, x]])

diagonal1_filtered_img = cv2.filter2D(img, -1, diagonal1_kernel)
plt.imshow(diagonal1_filtered_img, cmap='gray')
```

```python
x = -10
diagonal2_kernel = np.array([[-1, -1, x],
                            [-1, x, 1],
                            [x, 1, 1]])

diagonal2_filtered_img = cv2.filter2D(img, -1, diagonal2_kernel)
plt.imshow(diagonal2_filtered_img, cmap='gray')
```
