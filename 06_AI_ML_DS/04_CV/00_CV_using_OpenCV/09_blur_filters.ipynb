```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

img = cv2.imread("giraffe.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

# Average Blur

```python
kernel_size = 15
average_blur = cv2.blur(img, (kernel_size, kernel_size))
plt.imshow(average_blur)
```

# Gaussian Blur

```python
kernel_size = 27 # must be a odd value
std = 0
gaussian_blur = cv2.GaussianBlur(img, (kernel_size, kernel_size), std)
plt.imshow(gaussian_blur)
```

# Median Blur

```python
kernel_size = 15  # must be a odd value
median_blur = cv2.medianBlur(img, kernel_size)
plt.imshow(median_blur)
```
