```python
import matplotlib.pyplot as plt
import cv2

img = cv2.imread("shapes.png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

# Method 1 :- Through edge detection (Here canny)

```python
# 1. convert to grayscale
img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# 2. Apply edge detection or thresholding (Here edge detection)
gray = cv2.Canny(img_grey, 80, 150)
plt.imshow(gray, cmap='grey')
```

```python
# 3. Find contours
mode = cv2.RETR_TREE # hierarchy rel. between contours
method = cv2.CHAIN_APPROX_SIMPLE # contours points are stored
contours, hierarchy = cv2.findContours(gray, mode, method)
```

```python
# 4. Draw contours
contour_img = img.copy()
countour_idx = -1 # Parameter indicating a contour to draw, if its negative, all contours are drawn
color = (0, 255, 0) # green
thickness = 2
cv2.drawContours(contour_img, contours, countour_idx, color, thickness)
plt.imshow(cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB))
```

# Method 2 :- Through binary thresholding

```python
# 2. Apply edge detection or thresholding (Here Binary thresholding)
retval, binary = cv2.threshold(img_grey, 200, 255, cv2.THRESH_BINARY)
plt.imshow(binary, cmap='gray')
```

```python
# 3. Find contours
mode = cv2.RETR_TREE # hierarchy rel. between contours
method = cv2.CHAIN_APPROX_SIMPLE # contours points are stored
contours, hierarchy = cv2.findContours(binary, mode, method)
```

```python
# 4. Draw contours
contour_img = img.copy()
countour_idx = -1
color = (0, 255, 0) # green
thickness = 2
cv2.drawContours(contour_img, contours, countour_idx, color, thickness)
plt.imshow(cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB))
```

# Method 3 :- Through Range thresholding

```python
# 2. Apply edge detection or thresholding (Here Range thresholding)
import numpy as np
lower_bound = np.array([200], dtype=np.uint8)
upper_bound = np.array([255], dtype=np.uint8)

mask = cv2.inRange(img_grey, lower_bound, upper_bound)
plt.imshow(mask, cmap='gray')
```

```python
# 3. Find contours
mode = cv2.RETR_TREE # hierarchy rel. between contours
method = cv2.CHAIN_APPROX_SIMPLE # contours points are stored
contours, hierarchy = cv2.findContours(mask, mode, method)
```

```python
# 4. Draw contours
contour_img = img.copy()
countour_idx = -1
color = (0, 255, 0) # green
thickness = 2
cv2.drawContours(contour_img, contours, countour_idx, color, thickness)
plt.imshow(cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB))
```

```python
# 5. Draw Bounding box
contour_img = img.copy()
for cnt in contours:
  x, y, w, h = cv2.boundingRect(cnt)
  cv2.rectangle(contour_img, (x,y), (x+w, y+h), (0, 0, 255), 2)
plt.imshow(cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB))
```
